```python
import wandb
import pandas as pd

api = wandb.Api()
projects = [
    "ny0g507h3p_/woodbench-taskB",
    "ny0g507h3p_/woodbench-taskB-clean",
]

rows = []

for project in projects:
    for run in api.runs(project):
        if "90rotation" not in run.tags:
            continue
        if "clean" in run.name.lower():
            continue

        rows.append(
            {
                "project": project.split("/")[-1],
                "model": run.name,
                "val_acc": run.summary.get("val/acc"),
                "val_loss": run.summary.get("val/loss"),
                "test_acc": run.summary.get("test/accuracy"),
                "test_f1": run.summary.get("test/f1_macro"),
                "test_precision": run.summary.get("test/precision_macro"),
                "test_recall": run.summary.get("test/recall_macro"),
                "best_epoch": run.summary.get("best_epoch"),
            }
        )

df = pd.DataFrame(rows)
df = df.sort_values(by=["model", "project"]).reset_index(drop=True)
print(df)
```
