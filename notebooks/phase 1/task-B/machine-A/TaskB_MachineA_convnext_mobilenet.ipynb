## 4. Training Helpers — Parallel Runner

```python
import wandb, torch, threading, queue, time
from src.models import create_model
from src.train import TrainingEngine, create_optimizer, create_scheduler, create_loss_function
from src.utils import set_seed

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print(f"[Device] {DEVICE}")
if torch.cuda.is_available():
    print(f"[GPU] {torch.cuda.get_device_name(0)}")
    print(f"[VRAM] {torch.cuda.get_device_properties(0).total_mem / 1e9:.1f} GB")
```

```python
def train_one_in_thread(
    backbone: str,
    seed: int,
    data_root: str,
    batch_size: int,
    epochs: int,
    result_queue: queue.Queue,
    lock: threading.Lock,
    label: str,
):
    """Train a single model and push result to queue when done.
    All models share the same GPU via sequential GPU kernel calls;
    PyTorch releases VRAM between backward() passes, so two
    concurrent threads fit comfortably on 1× RTX 4090 24GB
    at batch_size=16 each (total ~13 GB)."""
    set_seed(seed)
    output_dir = ROOT / "experiments" / "taskB" / backbone / f"seed_{seed}"
    output_dir.mkdir(parents=True, exist_ok=True)

    train_loader, val_loader, _ = get_dataloaders(
        data_root=data_root,
        batch_size=batch_size,
        num_workers=4,
        task="taskB",
    )
    num_classes = len(train_loader.dataset.classes)

    with lock:
        print(f"[{label}] {backbone} s{seed} | {len(train_loader.dataset)} train | batch={batch_size}")

    # wandb: one run per model — no name collision
    wandb_run = wandb.init(
        project="woodbench-taskB",
        name=f"{backbone}_s{seed}_machine-A",
        config={
            "backbone": backbone,
            "seed": seed,
            "epochs": epochs,
            "batch_size": batch_size,
            "num_classes": num_classes,
            "lr_head": 1e-4,
            "lr_backbone": 1e-5,
            "label_smoothing": 0.1,
            "scheduler": "cosine",
            "amp": True,
            "machine": "A",
            "parallel": True,
        },
        dir=str(output_dir),
    )

    model = create_model(backbone, num_classes=num_classes,
                        pretrained=True, freeze_backbone=False)
    optimizer = create_optimizer(model, lr=1e-4, weight_decay=1e-4, backbone_lr=1e-5)
    scheduler = create_scheduler(optimizer, epochs=epochs, scheduler_type="cosine")
    criterion = create_loss_function("label_smoothing", label_smoothing=0.1)

    engine = TrainingEngine(
        model=model,
        train_loader=train_loader,
        val_loader=val_loader,
        optimizer=optimizer,
        scheduler=scheduler,
        criterion=criterion,
        device=DEVICE,
        use_amp=True,
    )

    result = engine.fit(
        epochs=epochs,
        output_dir=output_dir,
        name=f"{backbone}_seed{seed}",
        save_best_only=False,
    )

    wandb_run.finish()

    with lock:
        print(f"[{label}] {backbone} s{seed} DONE — val_acc={result['best_val_acc']:.2f}%")

    result_queue.put({"backbone": backbone, "seed": seed, "result": result})
```

```python
def run_parallel_pair(
    backbone_a: str, seed_a: int,
    backbone_b: str, seed_b: int,
    data_root: str,
    batch_a: int = 16,
    batch_b: int = 16,
    epochs: int = 50,
):
    """Launch 2 models in parallel threads (same GPU), wait for both."""
    print(f"\n{'='*65}")
    print(f"PARALLEL: {backbone_a} s{seed_a}  ||  {backbone_b} s{seed_b}")
    print(f"         batch {batch_a}            batch {batch_b}")
    print(f"{'='*65}")

    result_queue = queue.Queue()
    lock = threading.Lock()
    threads = []

    t_a = threading.Thread(
        target=train_one_in_thread,
        args=(backbone_a, seed_a, data_root, batch_a, epochs,
              result_queue, lock, "GPU-A"),
        name=f"{backbone_a}_s{seed_a}",
    )
    t_b = threading.Thread(
        target=train_one_in_thread,
        args=(backbone_b, seed_b, data_root, batch_b, epochs,
              result_queue, lock, "GPU-B"),
        name=f"{backbone_b}_s{seed_b}",
    )

    t_a.start()
    t_b.start()

    t_a.join()
    t_b.join()

    results = [result_queue.get(), result_queue.get()]
    print(f"\n[PAIR DONE] {backbone_a} s{seed_a} + {backbone_b} s{seed_b} both finished.")
    return results
```

## 5. Run All Seeds — ConvNeXt || MobileNet

Pairs: (ConvNeXt_s42, MobileNet_s42), (s123), (s456)
batch ConvNeXt=16, MobileNet=16

```python
SEEDS = [42, 123, 456]
all_results = []

for seed in SEEDS:
    pair_results = run_parallel_pair(
        backbone_a="convnext_tiny",    seed_a=seed,
        backbone_b="mobilenetv3_large", seed_b=seed,
        data_root=str(DATA_ROOT),
        batch_a=16,
        batch_b=16,
        epochs=50,
    )
    all_results.extend(pair_results)

print("\n" + "="*65)
print("MACHINE A — PARALLEL RUN COMPLETE")
print("="*65)
for r in all_results:
    print(f"  {r['backbone']} s{r['seed']}: {r['result']['best_val_acc']:.2f}%")
```

## 6. Summary

```python
import numpy as np

from collections import defaultdict
by_model = defaultdict(list)
for r in all_results:
    by_model[r["backbone"]].append(r["result"]["best_val_acc"])

print("="*60)
print("MACHINE A — FINAL SUMMARY")
print("="*60)
for model, accs in by_model.items():
    print(f"\n{model}:")
    for seed, acc in zip(SEEDS, accs):
        print(f"  seed={seed}: {acc:.2f}%")
    print(f"  Mean: {np.mean(accs):.2f}% ± {np.std(accs):.2f}%")

print(f"\n[wandb] All runs → woodbench-taskB (tag machine=A)")
print(f"[Next] Machine B: Swin + ResNet (parallel)")
print(f"[Next] Then: EfficientNet (standalone, batch 32)")
```
